## Setup

### Create `game_field`.

```python
# Initialize numpy.
import numpy as np

# Create our cell types.
cell = {
    "empty": "⬜",
    "baby": "🐣",
    "live": "🐔",
    "dead": "🍖"
}


# Define the game field size (columns, rows).
columns = 20
rows = 20
game_field_size = (columns, rows)
cells = columns * rows

# Create the game field as a numpy array filled with empty cells.
game_field = np.full(game_field_size, cell["empty"], dtype=object)

# Cleanup.
del game_field_size
```

## Create a random seeding.

```python
# Define probability of a cell being alive (between 0 and 1).
# Note: this probability of being alive is calculated for each cell independently, so the game field is not guaranteed to be populated at the given probability.
probability_of_alive = 0.20

# Track living cells and density.
living_cells = 0
living_density = 0.0

# Single line function to return a random boolean based on a given probability.
# This function returns True with the specified probability and False otherwise.
def random_probability(probability: float) -> bool: return np.random.rand() < probability

# Loop through each row of the field.
for row in range(game_field.shape[0]):

    # Loop through each column of the row.
    for col in range(game_field.shape[1]):

        # If the random probability is met, set the cell to live.
        if random_probability(probability_of_alive):
            game_field[row, col] = cell["baby"]
        else:
            game_field[row, col] = cell["empty"]
```

## Create a density monitoring function.

```python
def calculate_densities():
    
    # Initialize counters for each cell type.
    empty_cells = 0
    baby_cells = 0
    live_cells = 0
    dead_cells = 0

    # Initialize densities for each cell type.
    empty_cell_density = 0.0
    baby_cell_density = 0.0
    live_cell_density = 0.0
    dead_cell_density = 0.0

    # Loop through each row of the field.
    for row in range(game_field.shape[0]):

        # Loop through each column of the row.
        for col in range(game_field.shape[1]):

            # Count the number of each type of cell.
            if game_field[row, col] == cell["empty"]:
                empty_cells += 1
            elif game_field[row, col] == cell["baby"]:
                baby_cells += 1
            elif game_field[row, col] == cell["live"]:
                live_cells += 1
            elif game_field[row, col] == cell["dead"]:
                dead_cells += 1

    # Calculate the densities for each cell type.
    empty_cell_density = empty_cells / cells
    baby_cell_density = baby_cells / cells
    live_cell_density = live_cells / cells
    dead_cell_density = dead_cells / cells

    # Return the densities as a dictionary.
    return {
        "empty": empty_cell_density,
        "baby": baby_cell_density,
        "live": live_cell_density,
        "dead": dead_cell_density
    }

# Do a initial density calculation.
densities = calculate_densities()
# Print the initial game field and densities.
print("Initial Game Field:")
print(f"empty: {densities['empty']:.2f}, baby: {densities['baby']:.2f}, live: {densities['live']:.2f}, dead: {densities['dead']:.2f}")
```

## Create a function for printing the game_field.

```python
def print_game_field():
    for row in game_field:
        print(" ".join(row))
    print()

# Print the initial game field.
print_game_field()
```

## Create a function for checking a cells neighbors.

```python
def check_neighbors(row_add: int, col_add: int) -> int:

    # Initialize a counter for live neighbors.
    live_neighbors = 0

    # Loop through the neighboring cells, row first.
    for row in range(row_add - 1, row_add + 2):

        # Loop through the neighboring cells, column second.
        for col in range(col_add - 1, col_add + 2):
            
            # Skip the cell itself.
            if (row == row_add and col == col_add):
                continue

            # Check if the neighbor is within bounds, this will prevent index errors.
            if 0 <= row < game_field.shape[0] and 0 <= col < game_field.shape[1]:
                
                # Check if the neighbor is a live or baby cell.
                if game_field[row, col] == cell["live"] or game_field[row, col] == cell["baby"]:
                    
                    # Increment the live neighbors counter.
                    live_neighbors += 1

    # Return the count of live neighbors.
    return live_neighbors

# Edge case checks, to ensure we don't go out of bounds.
#    _ = check_neighbors(0,0)
#    _ = check_neighbors(columns - 1, 0)
#    _ = check_neighbors(0, rows - 1)
#    _ = check_neighbors(columns - 1, rows - 1)
```

## Create a function to iterate one frame of the game field.

```python
def iterate_game_field():

    # Declare the global game_field variable to modify it.
    global game_field
    
    # Print the game field before the iteration.
    print_game_field()
    
    # Copy the game field to last_game_field for tracking changes.
    last_game_field = game_field.copy()

    # Loop through each row of the field.
    for row in range(last_game_field.shape[0]):

        # Loop through each column of the row.
        for col in range(last_game_field.shape[1]):

            # Get the current cell from the last game field.
            current_cell = last_game_field[row, col]
            cell_live = cell["live"]
            cell_baby = cell["baby"]

            # Check the number of live neighbors for the current cell.
            live_neighbors = check_neighbors(row, col)

            # If the cell is alive (or a baby cell), check the rules.
            if current_cell == cell["live"] or current_cell == cell["baby"]:

                # If the cell is a baby, it becomes a live (adult) cell this iteration.
                if current_cell == cell["baby"]: game_field[row, col] = cell["live"]

                # If the cell has 2 or 3 live neighbors, it stays alive.
                if live_neighbors == 2 or live_neighbors == 3: game_field[row, col] = cell["live"]
                
                # Otherwise, it dies.
                else: 
                    game_field[row, col] = cell["dead"]
            
            # If the cell is empty or dead, check the rules for birth.
            if current_cell == cell["empty"] or current_cell == cell["dead"]:

                # If the cell has exactly 3 live neighbors, it becomes a baby cell.
                if live_neighbors == 3: game_field[row, col] = cell["baby"]
                
                # Otherwise, it remains empty, if its a dead cell, it turns empty.
                else: game_field[row, col] = cell["empty"]

    # Calculate the new densities after the iteration.
    new_densities = calculate_densities()

    # Return the new densities.
    return new_densities

```

## Create a function to render the emoji array as a image.

```python
from PIL import Image, ImageDraw, ImageFont

def render_game_field(matrix):

    # Define the size of each cell in the image (square) and alignment compensation.
    cell_size = 20
    left_border = 5
    top_border = 10
    right_border = 10
    bottom_border = 0
    
    # Get the number of rows and columns in the matrix.
    rows, cols = matrix.shape

    # Calculate the width and height of the image based on the number of cells and cell size.
    img_width = cols * cell_size + left_border + right_border
    img_height = rows * cell_size + top_border + bottom_border

    # Create white background image.
    img = Image.new('RGB', (img_width, img_height), color='white')
    draw = ImageDraw.Draw(img)

    # Load emoji-supporting font.
    font = ImageFont.truetype("seguiemj.ttf", size=int(cell_size * 0.8))

    # First loop through each row.
    for i in range(rows):

        # Then loop through each column in the row.
        for j in range(cols):
            
            # Get the emoji for the current cell.
            emoji = matrix[i, j]

            # Calculate the position to draw the emoji, compensating for borders.
            x = j * cell_size + left_border
            y = i * cell_size + top_border

            # Draw the emoji at the calculated position.
            if emoji == cell["live"]:
                # Draw live cell emoji.
                draw.text((x, y), emoji, font=font, fill='orange')
            elif emoji == cell["baby"]:
                # Draw baby cell emoji.
                draw.text((x, y), emoji, font=font, fill='green')
            elif emoji == cell["dead"]:
                # Draw dead cell emoji.
                draw.text((x, y), emoji, font=font, fill='red')
            else: draw.text((x, y), " ", font=font, fill='black')

    return img
```

## Iterate the game field.

```python
import os
image_array = []
for i in range(100):
    print(f"Iteration {i}:")
    image_array.append(render_game_field(game_field))
    
    # Iterate the game field.
    densities = iterate_game_field()

    # Print the densities after the iteration.
    print(f"empty: {densities['empty']:.2f}, baby: {densities['baby']:.2f}, live: {densities['live']:.2f}, dead: {densities['dead']:.2f}")

    # Print the game field after the iteration.
    output_path = os.path.join(os.getcwd(), "output.gif")

# Save as a gif
image_array[0].save(output_path, save_all=True, append_images=image_array[1:], duration=250, loop=1)
```
